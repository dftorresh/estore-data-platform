```python
bronze_warehouses_path = f"{BRONZE_PATH}/warehouses"
```

```python
df_warehouses_data = spark.read.format("jdbc") \
    .options(
        url = jdbc_url,
        user = jdbc_user,
        password = jdbc_password,
        dbtable = 'dbo.Warehouses'
    ) \
    .load()
```

```python
display(df_warehouses_data)
```

```python
df_warehouses_data.write.format("delta").mode("overwrite").save(bronze_warehouses_path)
```
